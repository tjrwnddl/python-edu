```python
import json
import requests

# API 호출 (서울시 공공데이터 나의 계정에서 api 인증키 삽입했음.)
response = requests.get('http://openapi.seoul.go.kr:8088/6b49584d44746a7238316f69786156/json/bikeStationMaster/1/5/')
json_data = response.json()

# 데이터 추출 및 처리
for item in json_data['bikeStationMaster']['row']:
    print(f"RNTLS_ID: {item['RNTLS_ID']}, ADDR1: {item['ADDR1']}, LAT: {item['LAT']}")

# 데이터 필터링 (예: LAT가 37.5보다 큰 항목)
filtered_data = [item for item in json_data['bikeStationMaster']['row'] if item['LAT'] > 37.5]

# 결과 활용
print(f"Number of items with LAT > 37.5: {len(filtered_data)}")
```
